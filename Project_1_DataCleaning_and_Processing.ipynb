```python
duplicate_rows = df.duplicated().sum()

print("\nDuplicate Rows Before Removal:", duplicate_rows)

df = df.drop_duplicates()

print("Duplicate Rows After Removal:", df.duplicated().sum())
```

```python
possible_id_columns = [
    "ID",
    "Id",
    "id",
    "Customer ID",
    "CustomerID",
    "Employee ID",
    "Order ID",
    "Transaction ID"
]

for col in possible_id_columns:

    if col in df.columns:

        print(f"\nChecking Duplicate IDs in '{col}'")

        print("Duplicate IDs:", df[col].duplicated().sum())

        df = df.drop_duplicates(subset=col)

        print("Duplicate IDs After Cleaning:",
              df[col].duplicated().sum())
```

```python
print("\nVerification Report")

print("-"*40)

print("Missing Values Remaining:",
      df.isnull().sum().sum())

print("Duplicate Rows Remaining:",
      df.duplicated().sum())

for col in df.columns:

    if "date" in col.lower():

        print(f"{col} Invalid Dates:",
              df[col].isna().sum())

# Define possible_id_columns before using it
possible_id_columns = ['OrderID', 'CustomerID', 'TrackingNumber']

for col in possible_id_columns:

    if col in df.columns:

        print(f"{col} Duplicate IDs:",
              df[col].duplicated().sum())
```

```python
output_file = "Cleaned_Dataset.xlsx"

df.to_excel(output_file, index=False)

print("\nCleaned dataset saved as:")
print(output_file)
```
